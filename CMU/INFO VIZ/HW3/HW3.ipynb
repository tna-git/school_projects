```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

## 1

```python
# Load the fmri dataset
fmri_data = sns.load_dataset("fmri")
fmri_data_parietal = fmri_data[fmri_data["region"] == "parietal"]
fmri_data_parietal
```

```python
sns.set_theme()
sns.relplot(
    data=fmri_data_parietal,
    x="timepoint", y="signal",
    col = "event",
    units = "subject",
    height=4, aspect=.7, kind="line", legend = False, estimator = None, linewidth=2)
plt.suptitle("Parietal Activity by region, All subjects")
```

## 2

```python
df = sns.load_dataset("diamonds")
df = df.sample(frac = 0.1, random_state = 42)
df
```

```python
sns.set_theme()
sns.relplot(
    data=df,
    x="carat", y="price",
    hue = "color",
    size = "carat", kind = "scatter")
```

## 3

```python
sns.set_theme()
sns.catplot(
    data=df,
    x="clarity", 
    y="carat",
    col="cut",
    col_wrap=3,
    kind="boxen",
    height=3,
    aspect=1.2
)
```
