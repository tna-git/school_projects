```python
import traitlets
```

```python
class MyTraitletsObject(traitlets.HasTraits):
    name = traitlets.Unicode("unamed")
    age = traitlets.Int(0)
```

```python
mytobj = MyTraitletsObject()
mytobj.age = 47
mytobj.name = 'Bob'
```

```python
print("Hello my name is", mytobj.name, ". I am", mytobj.age)
```

```python
def name_change(change):
    print(change)
    
mytobj.observe(name_change, ['name'])
```

```python
mytobj.name = 'Linda'
```

```python
def name_change(change):
    print("Hello my name is", change['owner'].name, ". I am", change['owner'].age)

mytobj.observe(name_change, ['name'])
```

```python
mytobj.name = 'Trishla'
```

```python
mytobj.unobserve(name_change, ['name'])
```
