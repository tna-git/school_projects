```python
import pandas as pd
import bqplot
import numpy as np
import traitlets
import ipywidgets
import matplotlib.pyplot as plt
```

```python
license = pd.read_csv("https://raw.githubusercontent.com/UIUC-iSchool-DataViz/is445_bcubcg_fall2022/main/data/licenses_fall2022.csv")
```

```python
license['Original Issue Date'] = pd.to_datetime(license["Original Issue Date"],
                                           format='%m/%d/%Y', 
                                           errors='coerce')
license['Expiration Date'] = pd.to_datetime(license["Expiration Date"],
                                           format='%m/%d/%Y', 
                                           errors='coerce')
```

```python
license['days between Original Issue Date and Expiration Date'] = (license['Expiration Date'] - license['Original Issue Date']).dt.days
license['days between Original Issue Date and Expiration Date']
license
```
