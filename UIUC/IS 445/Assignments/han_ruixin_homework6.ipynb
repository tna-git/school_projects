```python
import pandas as pd
import bqplot
import numpy as np
import traitlets
import ipywidgets
import matplotlib.pyplot as plt
```

```python
license = pd.read_csv("https://raw.githubusercontent.com/UIUC-iSchool-DataViz/is445_bcubcg_fall2022/main/data/licenses_fall2022.csv")
```

```python
license['Original Issue Date'] = pd.to_datetime(license["Original Issue Date"],
                                           format='%m/%d/%Y', 
                                           errors='coerce')
license['Expiration Date'] = pd.to_datetime(license["Expiration Date"],
                                           format='%m/%d/%Y', 
                                           errors='coerce')
```

```python
license['days between Original Issue Date and Expiration Date'] = (license['Expiration Date'] - license['Original Issue Date']).dt.days
license['days between Original Issue Date and Expiration Date']
```

```python
# Barplot/histogram
x_scl = bqplot.LinearScale() # note we are back to linears
y_scl = bqplot.LinearScale()
ax_xcl = bqplot.Axis(label='Year', scale=x_scl)
ax_ycl = bqplot.Axis(label='Median number of days between Original Issue Date and Expiration Date', scale=y_scl, 
                    orientation='vertical', side='right')

# hist = bqplot.Hist(sample = data3d[0,0,:],
#                    normalized = False, # normalized=False means we get counts in each bin
#                    scales = {'sample': x_sch, 'count': y_sch}, # sample is data values, count is frequency
#                    bins = 5) # number of bins
```

```python
license['Original Issue Year'] = license['Original Issue Date'].dt.year
license['Expiration Year'] = license['Expiration Date'].dt.year
license['Year'] = license.groupby(['Original Issue Year'])['Expiration Year'].transform('count')
license['Year']
license
```

```python
license['Original Issue Year']
```

```python
license_Year = pd.pivot_table(license, values='days between Original Issue Date and Expiration Date', index=['Original Issue Year'], columns=['Expiration Year'],
                      aggfunc=np.median, fill_value=0)
license_Year
```

```python
license_Year=license.groupby('Original Issue Year')['days between Original Issue Date and Expiration Date'].median().reset_index()
license_Year
```

```python
import bqplot.pyplot as plt
from bqplot import Tooltip
fig_bar = plt.figure(title="Median Days between Original Issue Date and Expiration Date")
bar_chart = plt.bar(x = license_Year['Original Issue Year'], y = license_Year['days between Original Issue Date and Expiration Date'])
bar_chart.colors = ["green"]
bar_chart.tooltip = Tooltip(fields=["x", "y"], labels=["Original Issue Year", "Median days"])
plt.xlabel("Original Issue Year")
plt.ylabel("Median days")

plt.show()
```

```python
table = pd.pivot_table(license, values= 'days between Original Issue Date and Expiration Date', index= ['County'] , columns= ['License Type'] ,
                      aggfunc=np.mean, fill_value = 0)
table
```

```python
# HEATMAP
# 2. scales
col_sc = bqplot.ColorScale(scheme='Blues')# <3
x_sc = bqplot.OrdinalScale() # ordered bins
y_sc = bqplot.OrdinalScale()

# 3. axis
col_ax = bqplot.ColorAxis(scale=col_sc, label='Mean', orientation='vertical',
                         side='right')
x_ax = bqplot.Axis(scale=x_sc,label='License Type',tick_rotate=90)
y_ax = bqplot.Axis(scale=y_sc, label='County', orientation='vertical')

# 4. Mark - heatmap
heat_map = bqplot.GridHeatMap(color=table, 
                              row=table.index.values.astype('str'),
                              column=table.columns.values.astype('str'),
                              scales={'color':col_sc,
                                                 'row':y_sc,
                                                 'column':x_sc},
                             interactions={'click':'select'},
                             selected_style={'fill':'blue'})

# 5. interactions
def get_data_value(change):
    if len(change['owner'].selected) == 1: #only 1 selected
        i,j = change['owner'].selected[0]
        print(i,j)
heat_map.observe(get_data_value, 'selected')


fig = bqplot.Figure(marks=[heat_map], axes=[col_ax, x_ax, y_ax])
fig
```

```python
# HEATMAP
# 2. scales
col_sc = bqplot.ColorScale(scheme='Blues')# <3
x_sc = bqplot.OrdinalScale() # ordered bins
y_sc = bqplot.OrdinalScale()

# 3. axis
col_ax = bqplot.ColorAxis(scale=col_sc, label='Mean', orientation='vertical',
                         side='right')
x_ax = bqplot.Axis(scale=x_sc,label='License Type',tick_rotate=90)
y_ax = bqplot.Axis(scale=y_sc, label='County', orientation='vertical')

# 4. Mark - heatmap
heat_map = bqplot.GridHeatMap(color=table, 
                              row=table.index.values.astype('str'),
                              column=table.columns.values.astype('str'),
                              scales={'color':col_sc,
                                                 'row':y_sc,
                                                 'column':x_sc},
                             interactions={'click':'select'},
                             selected_style={'fill':'blue'})

# 5. interactions

fig = bqplot.Figure(marks=[heat_map], axes=[col_ax, x_ax, y_ax])
fig

def get_data_value(change):
    if len(change['owner'].selected) == 1: #only 1 selected
        i,j = change['owner'].selected[0]
        county_name=table.index.values.astype('str')[i]
        license_type=table.columns.values.astype('str')[j]
        region_mask = ( (license['County']==county_name) &\
              (license['License Type']==license_type))
        data=license.loc[region_mask]
        data_y=data.groupby('Original Issue Year')['days between Original Issue Date and Expiration Date'].median()
        bar.x = data['Original Issue Year']
        bar.y = data_y

        
        
heat_map.observe(get_data_value, 'selected')
```

```python
fig = bqplot.Figure(marks = [heat_map], axes = [col_ax, y_ax, x_ax])
fig_bar = bqplot.Figure(marks = [bar], axes = [ax_xcl, ax_ycl])
```
