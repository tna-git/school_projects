```python
columns = {'DMG_LEVEL': map3['DMG_LEVEL'], 'STRUCTTYPE': map2['STRUCTTYPE'], 'none': ''}
def map_choice(choice1, choice2,columns):
    figure, ax = plt.subplots(figsize=(10,10))
    choice1.to_crs(choice2.crs).plot(ax=ax)
    choice2.plot(ax = ax, column = columns, cmap = "viridis", alpha = 0.5, legend = True)   
```

```python
map_choice(hmap, map2, 'STRUCTTYPE')
```

```python
choose = [1, 2, 3]

@ipywidgets.interact(choose = choose, color_map = plt.colormaps())
def map_viz(choose, color_map):
    if choose == 1:
        fig, ax = plt.subplots(figsize=(10,10))
        hmap.to_crs(map1.crs).plot(ax=ax)
        map1.plot(ax=ax, cmap = color_map)
        plt.xlabel("Latitude")
        plt.ylabel("Longitutde")
        plt.title("Drainage Areas of Houston")
    elif choose == 2:
        new_fig, ax = plt.subplots(figsize=(10,10))
        map1.to_crs(map2.crs).plot(ax=ax)
        map2.plot(ax = ax, column = 'STRUCTTYPE', cmap = color_map, alpha = 0.5, legend = True)
        plt.xlabel("Latitude")
        plt.ylabel("Longitutde")
        plt.title("Stormdrain Network Structures of Houston")
    else:
        har, ax = plt.subplots(figsize=(10,10))
        map1.to_crs(map3.crs).plot(ax=ax)
        map3.plot(ax = ax, column = "DMG_LEVEL", cmap = color_map, markersize = 9,alpha = 0.5, legend = True)
        plt.xlabel("Latitude")
        plt.ylabel("Longitutde")
        plt.title("Areas Affected by Hurricane Harvey (shown in terms of Damage Level) in Houston")
```
