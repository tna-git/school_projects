```python
df = df.fillna(0)
df = pd.read_csv('https://raw.githubusercontent.com/UIUC-iSchool-DataViz/is445_bcubcg_fall2022/main/data/licenses_fall2022.csv', na_values = ['None'])
df
```

```python
df['Original Issue Date'] = pd.to_datetime(df['Original Issue Date'],errors='coerce',format='%m/%d/%Y')
df['Original Issue Year'] = df['Original Issue Date'].dt.year

df['Expiration Date'] = pd.to_datetime(df['Expiration Date'],errors='coerce',format='%m/%d/%Y')
df['Expiration Year'] = df['Expiration Date'].dt.year


df['Difference_Days'] = df['Expiration Date'] - df['Original Issue Date']
df['Difference_Days'] = df['Difference_Days'].dt.days
df['Difference_Days']
```

```python
table = pd.pivot_table(df, values= 'Difference_Days', index= 'County' , columns= 'License Type' ,
                      aggfunc=np.mean, fill_value = 0)
table
```

```python
# (I) For the heatmap
# add scales - colors, x & y
col_sc = bqplot.ColorScale(scheme="RdPu")
x_sc = bqplot.OrdinalScale()
y_sc = bqplot.OrdinalScale()

# create axis - for colors, x & y
c_ax = bqplot.ColorAxis(scale = col_sc, 
                        orientation = 'vertical', 
                        side = 'right')
x_ax = bqplot.Axis(scale = x_sc, label='License Type Alphabetized (Numbered 1 - 26)')
y_ax = bqplot.Axis(scale = y_sc, 
                   orientation = 'vertical', 
                   label = 'County Alphabetized and Numbered')

heat_map = bqplot.GridHeatMap(color = table,
                              scales = {'color': col_sc,
                                        'row': y_sc,
                                        'column': x_sc},
                              interactions = {'click': 'select'},
                              anchor_style = {'fill':'blue'}, 
                              selected_style = {'opacity': 1.0},
                              unselected_style = {'opacity': 1.0})

fig = bqplot.Figure(marks = [heat_map], axes = [c_ax, y_ax, x_ax])
fig
```

```python
# (II) Bar plot
# scales & ax in usual way
x_scl = bqplot.OrdinalScale() # note we are back to linears
y_scl = bqplot.OrdinalScale()
ax_xcl = bqplot.Axis(label='Year', scale=x_scl)
ax_ycl = bqplot.Axis(label='Median days', scale=y_scl, 
                    orientation='vertical', side='left')
# for the lineplot of duration in a region as a function of year
# lets start with a default region & year
i,j = 0,0
ey = [df['Expiration Year'][i], df['Expiration Year'][i+1]]
oiy = [df['Original Issue Year'][j],df['Original Issue Year'][j+1]]
region_mask = ( (df['Original Issue Year'] >= oiy[0]) & (df['Original Issue Year']<=oiy[1]) &\
                (df['Expiration Year'] >= ey[0]) & (df['Expiration Year']<=ey[1]) )

# we can see this selects for the upper right point of our heatmap
oiy, ey, df['Original Issue Year'][region_mask]

# lets plot the durations as a function of year there
dur, dur_edges = np.histogram(df['Expiration Year'][region_mask],
                              weights=df['Difference_Days'][region_mask],
                              bins=10)
# like before with our histograms
dur_centers = [(dur_edges[i]+dur_edges[i+1])*0.5 for i in range(len(dur_edges)-1)]
# make histogram by hand, weighting by duration
duration_hist = bqplot.Bars(x=dur_centers, y=dur, 
                          scales={'x':x_scl, 'y':y_scl})
fig_dur = bqplot.Figure(marks = [duration_hist], axes = [ax_xcl, ax_ycl])
fig_dur
```
