### A. Plot a scatter plot of lung cancer deaths against cigarette sales, using the two letter abbreviation for each state as a marker. You should see two fairly obvious outliers. The backstory at http://lib.stat.cmu.edu/DASL/Stories/cigcancer.html suggests that the unusual sales in Nevada are generated by tourism (tourists go home, and die there) and the unusual sales in DC are generated by commuting workers (who also die at home).

```python
df = pd.read_csv("cigcancer-1.txt", delimiter = '\t')
df
```

```python
lung = df['LUNG'].to_list()
cig = df['CIG'].to_list()
state = df['STATE'].to_list()

for i in range(len(lung)):
    plt.scatter(lung[i], cig[i], marker = f'${state[i]}$', alpha = 1.0)
plt.show()
```

### B.What is the correlation coefficient between per capita cigarette sales and lung cancer deaths per 100K population? Compute this with, and without the outliers. What effect did the outliers have? Why?

```python
with_outliers = np.corrcoef(cig, lung)
with_outliers
```

```python
df_o = df.sort_values(by = 'CIG')
df_o = df_o.head(42)
without_outliers = np.corrcoef(df_o['CIG'], df_o['LUNG'])
without_outliers
```

The outliers, when removed, increased the coefficient. Since they were higher values compared to the res of the data, they altered the linearity of thye data.

## 5. Download the daily adjusted closing stock prices for the past 3 months of the Coca-Cola Company and PepsiCo

```python
pep = pd.read_csv("PEP.csv")
pep
```

```python
ko = pd.read_csv("KO.csv")
ko
```

### A. Use this data to find the correlation coefficient between the stock prices of these two corporations

```python
np.corrcoef(ko['Adj Close'],pep['Adj Close'])
```

### B. Plot a scatter plot with KO prices on the horizontal axis and PEP prices on the vertical axis

```python
plt.plot(ko['Adj Close'],pep['Adj Close'], 'o')
plt.xlabel("Coca-Cola Adjusted Closing Stock Prices")
plt.ylabel("Pepsi Adjusted Closing Stock Prices")
plt.title("Coca Cola Stocks vs Pepsi Stocks")
plt.show()
```

### C. Add a prediction line to your plot that shows predictions of PEP prices from KO prices

```python
plt.plot(ko['Adj Close'],pep['Adj Close'], 'o')
plt.xlabel("Coca-Cola Adjusted Closing Stock Prices")
plt.ylabel("Pepsi Adjusted Closing Stock Prices")
plt.title("Coca Cola Stocks vs Pepsi Stocks")
m, b = np.polyfit(ko['Adj Close'],pep['Adj Close'], 1)
plt.plot(ko['Adj Close'], m*ko['Adj Close'] + b)
```
