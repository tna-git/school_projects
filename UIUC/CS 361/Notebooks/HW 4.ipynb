```python
import pandas as pd
import numpy as np
from scipy.stats import bernoulli
from scipy.stats import binom
import matplotlib.pyplot as plt
```

An airline company runs a flight that has 10 seats. Each passenger who buys a ticket has a probability p of turning up for the flight. The gender of the passengers is not known until they turn up for a flight, and women buy tickets with the same frequency that men do. The pilot is eccentric, and will not fly unless at least two women turn up.

A. How many tickets should the airline sell to ensure that the expected number of passengers that turn up is greater than 10?

```python
for i in range(100):
    tickets = binom(p=0.5, n =i)
    if (tickets.expect() >= 10):
        print("Expected value:", tickets.expect(),"number of tickets:", i)
        break
```

So the plane needs to sell at least 21 tickets to ensure that the expected number of passengers that turn up is greater than 10.

B. The airline sells 10 tickets.What is the expected number of passengers on the aircraft, given that it flies? (i.e. that at least two women turn up). Estimate this value with a simulation.

```python
# what we know:
# 1. the plane flies, so at least 2 women show up, thus 2 seats are taken
# 2. The airline sells 10 tickets, 2 are bught by women.
# probability of a person getting on the plane given the 2 women is 0.5 (0.25/0.5) = 0.5
```

```python
p_vals = []
results = []
for i in range(1,11):
    p = i/10
    p_vals.append(p)
    results.append(np.random.binomial(10, p, 10000))
    plt.hist(results, bins=np.arange(10+2), edgecolor='black')
    plt.xlabel("Expected Number of people on the plane (besides the 2 women) when the plane flies")
    plt.ylabel("Frequency of Successes")
    plt.title("Binomial Simulation for when p is " + str(p))
    plt.legend()
    plt.show()
```
