```python
import pandas as pd
from sklearn.model_selection import cross_validate
from sklearn.linear_model import LinearRegression
import numpy as np
```

```python
df = pd.read_csv("ew.csv")
df
```

```python
# Separate features (X) and target (y)
X = df.drop(columns=['label'])  # Assuming 'label' is the target column
y = df['label']
```

```python
# Initialize the LinearRegression model
model = LinearRegression()
```

```python
# Set up 10-fold cross-validation
cv_results = cross_validate(model, X, y, cv=10, scoring='r2', return_train_score=True)
print(cv_results)
```

```python
# Calculate mean and standard deviation of R^2 for training and testing data
train_r2 = (cv_results['train_score'].sum())/10
train_r2_sd = cv_results['train_score'].std()

test_r2 = cv_results['test_score'].mean()
test_r2_sd = cv_results['test_score'].std()
```

```python
print(train_r2, train_r2_sd, test_r2, test_r2_sd)
```
