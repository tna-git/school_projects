```python
import pandas as pd
from sklearn.neighbors import KNeighborsRegressor
from sklearn.model_selection import cross_validate, KFold
```

```python
data = pd.read_csv('C:/Users/Trish/Documents/CS 307/df.csv')

# Step 2: Prepare the data
# Assuming the last column is the label and the rest are features
X = data.iloc[:, :-1]  # Features
y = data.iloc[:, -1]   # Label

# Step 3: Initialize the KNeighborsRegressor
knn_regressor = KNeighborsRegressor(n_neighbors=3)

# Step 4: Set up 10-fold cross-validation
kf = KFold(n_splits=10, shuffle=True)

# Step 5: Evaluate the model using cross_validate
cv_results = cross_validate(knn_regressor, X, y, cv=kf, scoring='r2', return_train_score=True)

# Step 6: Calculate the mean and standard deviation of R^2 scores
train_r2 = cv_results['train_score'].mean()
train_r2_sd = cv_results['train_score'].std()
test_r2 = cv_results['test_score'].mean()
test_r2_sd = cv_results['test_score'].std()

# Output the results
print(f"Train R²: {train_r2}, Train R² SD: {train_r2_sd}")
print(f"Test R²: {test_r2}, Test R² SD: {test_r2_sd}")
```
